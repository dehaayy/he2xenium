This is specifically usefull because you can build a pipeline from your data path using this simple python command to run `A SINGLE` file, you can assign this into other slurms in parallel


- Spatial Image Alignment (`run_alignment.py`)

Run this pipeline to co-register histology (H&E) slide images with Xenium spatial transcriptomics coordinates.

```bash
cd "/path/to/wherever_you_store_the_functions_f(x)"

python run_alignment.py \
  --he "/home/day/Desktop/deha_base/LIANG/FFPE Xenium H&E/FFPE Xenium Run #1/H&E images Xenium slides post-analysis/2018-50 & 2019-5383 Xenium slide.ndpi" \
  --annotation "/home/day/Desktop/deha_base/LIANG/FFPE Xenium H&E/FFPE Xenium Run #1/H&E images Xenium slides post-analysis/2018-50 & 2019-5383 Xenium slide.ndpi.ndpa" \
  --xenium "/home/day/Desktop/deha_base/LIANG/RESEG/2018-50_reseg/cells.parquet" \
  --auto-crop \
  --location 2 \
  --out "/home/day/Desktop/deha_base/auto_allignment/2018-50"

```

- CLI Arguments:

* `--he`: Path to the raw brightfield whole-slide image (`.ndpi`).
* `--annotation`: Path to the annotation file (`.ndpa`) defining tissue boundaries/ROIs.
* `--xenium`: Path to the Xenium cell coordinates table (`cells.parquet`).
* `--auto-crop`: Automatically crops the image to the relevant tissue bounding box.
* `--location`: Specific section/region index to process when multiple tissues exist on one slide.
* `--out`: Output folder where all generated files (transformed coordinates, transformation matrices, and QC overlays) are saved.

````python
df = pd.read_csv("/home/day/Desktop/PYTHON/Liu/my_allignment_clean_final/example_df.csv",index_col = 0)
per_instance = df.loc["2018-50"]
````

````python
OUT = f"/home/day/Desktop/deha_base/auto_allignment/{per_instance.name}"

IMAGE, annotations, ndpi_meta = mall.load_ndpi_with_annotations(
    per_instance['H&E_path'], per_instance['annotation_path'], level=5, return_meta=True)
xenium_obs = pd.read_parquet(per_instance['xenium_path']).set_index('cell_id')

if per_instance['auto_crop'] == True:
    val = per_instance.get("sample_count")
    count = int(val) if pd.notna(val) else None
    selected_sample_crop, sample_anns, info = ac.crop_sample_with_annotations(
        IMAGE, annotations, location=int(per_instance['location']),
        n_samples=count, pad_frac=0.05, plot=False, out_dir=OUT)
else:
    selected_sample_crop, sample_anns = IMAGE, annotations
    info = {"total_offset": (0, 0)}

binary_img, confidence_img = mall.kmeans_binary(selected_sample_crop, plot=False)
CROPPED_ORIGINAL_IMG, img_diag_size, crop_coordinates = mall.contour_crop_binary(
    binary_img, ratio=0.01, pad_frac=0.2, sigma=1.0, plot=False)

HE_CROPPED = mall.pad_crop(selected_sample_crop, crop_coordinates, value=255)
xenium_coord_df, xenium_image, meta = mall.xenium_binarize(
    xenium_obs, pad_frac=0.2, ratio=0.015, img_diag_size=img_diag_size, plot=False,
    column_names=["centroid_x", "centroid_y"])
````

````python
result = awg.run_genetic_algorithm(
    xenium_image,
    CROPPED_ORIGINAL_IMG,
    xenium_coord_df,
    population_size=1000,
    generations=20,

    elite_frac=0.03,
    immigrant_frac=0.08,
    mating_frac=0.65,

    tournament_size=3,

    mutation_rate_hi=0.40,
    mutation_rate_lo=0.08,

    sigma_hi=0.20,
    sigma_lo=0.02,

    diversity_floor=0.12,
    stagnation_limit=8,
    n_workers=None

)
````

````python
r = awg.apply_ga_result(result, xenium_image, CROPPED_ORIGINAL_IMG, xenium_coord_df)
awg.plot_chamfer_qc(CROPPED_ORIGINAL_IMG, xenium_coord_df, display_px=600, out_dir=OUT)

counts, crop_anns, offset, score = awg.label_cells_from_annotations(
    xenium_coord_df, sample_anns, binary_img, CROPPED_ORIGINAL_IMG,
    crop_coordinates=crop_coordinates, out_dir=OUT)

final_df = awg.to_original_he(xenium_coord_df, crop_coordinates, info["total_offset"],
                   level_downsample=ndpi_meta["downsample"], out_dir=OUT)
````
